```python
# Let's create a CSV (comma separated values) file
header = "Name,Age,Grade\n"
rows = [
    "Jaume,30,8.9\n",
    "Francisco,25,7.1\n",
    "Elena,35,9.2\n"
]
```

```python
with open("grades.csv", "w") as f:
    f.write(header) # Write the header
    
    # Write each row of data
    for row in rows:
        f.write(row)
```

```python
with open("grades.csv", "r") as f:
    lines = f.read().splitlines()
    
header = lines.pop(0)
header = header.split(',')

print(header)

grades = {'students': []}
# create dictionary
for line in lines:
    student_dict = {}
    values = line.split(',')
    for idx, column in enumerate(header):
        student_dict[column] = values[idx]
    grades['students'].append(student_dict)
    
grades
```

## Basic Libraries II

Similar to JSON, Python includes a Pickle library. However, in contrast to the JSON format, Pickle is a Python-specific serialization format. The Pickle library provides tools to serialize Python objects, which involves transforming them into a stream of bytes. It also allows you to read these byte streams by deserializing them, transforming them back into their original Python objects.

In contrast to the JSON format, the binary format is usually more compact and, therefore, more efficient.

```python
import numpy as np
data = np.random.rand(10)

import pickle

# Serializing (dumping) the object
with open('data.pkl', 'wb') as f:
    pickle.dump(data, f)

# Deserializing (loading) the object
with open('data.pkl', 'rb') as f:
    loaded_data = pickle.load(f)

print(loaded_data)
```

To work with **Parquet** files, you need either the **pyarrow** or **pandas** library. Parquet is a columnar storage format, meaning that each row represents a sample, and each column represents an attribute. This is a powerful format commonly used as a standard in platforms like **Hugging Face**.

```python
import pandas as pd # if it is not working, simply uncomment the following line
# !pip install pandas

# Creating a DataFrame
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35]
})

# Writing DataFrame to Parquet file with Pandas
df.to_parquet('data.parquet')

# Reading DataFrame from Parquet file with Pandas
df_loaded = pd.read_parquet('data.parquet')

print(df_loaded)
```

Let's now try to use them to order the annotations by date

```python
import re
import glob
import os
from datetime import datetime

# Regex pattern, r in front of strings tell python to treat them as raw strings
# we do this so slashes don't get interpret as scaping symbol
pattern = r'(\d{8})_(\d{6})_SN(\d+)_QUICKVIEW_VISUAL_([\d_]+)_([A-Za-z0-9\-_.]+)\.txt' 

annotations = glob.glob('session_4/annotations/*.txt')

# let's create a dictionary where per each annotations we gather the datetime object
ann_datetime = []

for annotation in annotations:

    # extract the file name
    filename = os.path.basename(annotation)
    
    # Search and extract values
    match = re.match(pattern, filename)
    if match:
        date, time, _, _, _ = match.groups()

        # Put them together, e.g. "20240101192856"
        datetime_str = date + time 

        # Parse the string into a datetime object
        datetime_obj = datetime.strptime(datetime_str, "%Y%m%d%H%M%S")

        # Output the datetime object
        print(f"Datetime Object: {datetime_obj}")
        
        ann_datetime.append((filename, datetime_obj))
```

```python
indices = np.argsort([date for name, date in ann_datetime])
indices
```

```python
for i in indices:
    print(ann_datetime[i][0])
```

```python
import json

# Ruta de los archivos de anotación (ajustar según sea necesario)
annotations = glob.glob('/Users/montserratmiroverdaguer/Documents/Esade /Master/Python/Assignments/session_4/annotations/*.txt')

# Inicializar un diccionario para almacenar listas de archivos por mes/año
annotations_by_month = {}

# Patrón de expresión regular para extraer 'YYYYMMDD' del nombre del archivo
date_pattern = re.compile(r'(\d{4})(\d{2})(\d{2})')  # Esto coincidirá con 'YYYYMMDD'

# Procesar cada archivo
for file in annotations:
    filename = os.path.basename(file)
    match = date_pattern.match(filename)
    
    if match:
        year, month, day = match.group(1), match.group(2), match.group(3)
        date_str = f"{year}-{month}-{day}"
        try:
            date = datetime.strptime(date_str, "%Y-%m-%d")
            
            # Convertir la tupla (year, month) en un formato de cadena "YYYY-MM" para JSON
            month_year_str = f"{date.year}-{date.month:02d}"
            
            # Inicializar lista si no existe el mes/año en el diccionario
            if month_year_str not in annotations_by_month:
                annotations_by_month[month_year_str] = []
            
            # Agregar solo el nombre del archivo a la lista
            annotations_by_month[month_year_str].append(filename)
        except ValueError:
            print(f"Skipping file {filename}: Invalid date format")

# Guardar en formato JSON
with open('annotations_by_month.json', 'w') as json_file:
    json.dump(annotations_by_month, json_file)

# Cargar y verificar el JSON
with open('annotations_by_month.json', 'r') as json_file:
    loaded_json_data = json.load(json_file)

# Mostrar resultados en formato vertical
print("JSON loaded successfully:")
for month_year, files in loaded_json_data.items():
    print(f"Month/Year: {month_year}")
    for file in files:
        print(f"  - {file}")
```

```python
import pickle

# Guardar el diccionario en formato Pickle
with open('annotations_by_month.pkl', 'wb') as pickle_file:
    pickle.dump(annotations_by_month, pickle_file)

# Cargar y verificar Pickle
with open('annotations_by_month.pkl', 'rb') as pickle_file:
    loaded_pickle_data = pickle.load(pickle_file)

# Mostrar resultados en formato vertical
print("Pickle loaded successfully:")
for month_year, files in loaded_pickle_data.items():
    print(f"Month/Year: {month_year}")
    for file in files:
        print(f"  - {file}")
```

```python
# Inicializar un nuevo diccionario para almacenar detalles de las anotaciones por mes/año
annotations_by_month_with_details = {}

for file in annotations:
    filename = os.path.basename(file)
    match = date_pattern.match(filename)
    
    if match:
        year, month, day = match.group(1), match.group(2), match.group(3)
        date_str = f"{year}-{month}-{day}"
        try:
            date = datetime.strptime(date_str, "%Y-%m-%d")
            month_year = (date.year, date.month)
            
            # Inicializar lista si no existe el mes/año en el diccionario
            if month_year not in annotations_by_month_with_details:
                annotations_by_month_with_details[month_year] = []
            
            # Agregar un diccionario con `name` y `date`
            annotations_by_month_with_details[month_year].append({
                'name': filename,
                'date': date
            })
        except ValueError:
            print(f"Skipping file {filename}: Invalid date format")

# Mostrar resultados en formato vertical
print("Annotations with details by month/year:")
for (year, month), annotations in annotations_by_month_with_details.items():
    print(f"Year: {year}, Month: {month}")
    for annotation in annotations:
        print(f"  - Name: {annotation['name']}, Date: {annotation['date'].strftime('%Y-%m-%d')}")
```

```python
# Filtrar anotaciones para el segundo semestre de 2024
second_half_2024 = []

for (year, month), annotations_list in annotations_by_month_with_details.items():
    if year == 2024 and 7 <= month <= 12:
        second_half_2024.extend(annotations_list)  # Agregar todas las anotaciones de estos meses

# Ordenar las anotaciones por fecha
sorted_annotations = sorted(second_half_2024, key=lambda x: x['date'])

# Imprimir las anotaciones ordenadas
print("Annotations from the second half of 2024, sorted by date:")
for annotation in sorted_annotations:
    print(f"Date: {annotation['date'].strftime('%Y-%m-%d')}, Name: {annotation['name']}")

# No obtenemos respuesta porque no hay datos de las anotaciones desde julio hasta diciembre. Como podemos ver en el primer ejercicio, solo estan desde enero hasta junio. En el caso que hubiera, como el codigo esta correcto saldrían. 
```
