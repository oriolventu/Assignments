## Pandas I

**Pandas** is one of the most used libraries within the Data Science community. It provides a full set of tools to work with:
- 2D data via DataFrame class
    - SQL/Spreadsheet-like datasets (tabular data)
    - Arbitrary matrix data with row and column labels

- 1D data via Series class

## Pandas I

And how do we know about missing elements and handle them?

```python
# What about missing values? Missing values are usually represented as NaN (Not a Number)
import numpy as np
data = {'Name': ['Amelie', 'Edgar', 'Carlos', 'Victor'],
        'Age': [24, 27, 22, 32],
        'Country': ['FR', np.nan, 'ES', 'GE']}
df = pd.DataFrame(data)
df
```

```python
# we can simply know which ones present missing values by using isna or isnull
df.isna() # at the whole dataframe level
```

```python
# and at the column level
df['Country'].isnull() # same as .isna()
```

```python
# Simple way of filling these ones is achieved by calling fillna
df.fillna('RE')
```

```python
# There's also possible thing you can do, use the mode (most common value)
#df['Country'].mode()
df.fillna(df['Country'].mode()[0])
```

```python
# IMPORTANT: if we check again the dataframe, we will see that the dataframe keeps having NaNs
# This is because the intented behaviour is usually to keep the DataFrame immutable
# If you want to actually modify you can pass ot fillna, inplace=True
df
```

```python
# A more drastic operation is to drop any column or row with missing values using
# here, we will modify the original dataframe and drop the rows and cols with nans
df.at[0, 'Age'] = np.nan
df
```

```python
df.dropna()
```

## Pandas I

One of the handy tools you have in **Pandas** is **groupby**. It allows you to group by any unique attributes within a column and get statistics:
- count
- mean
- sum
- min and max
- multiple aggregations
- group using multiple columns

```python
data = {
    'Name': ['Amelie', 'Edgar', 'Carlos', 'Victor', 'Sofia', 'Jin', 
             'Marta', 'Ali', 'Emily', 'Ravi', 'Chen', 'Fatima', 'Saham'],
    'Age': [24, 27, 22, 32, 29, 31, 28, 26, 23, 34, 25, 30, 26],
    'Country': ['FR', 'US', 'ES', 'GE', 'PT', 'KR',
                'ES', 'AE', 'US', 'IN', 'CN', 'AE', 'AE'],
    'Salary': [70000, 110000, 65000, 82000, 48000, 39000, 45000,
            90000, 97000, 31000, 49000, 85000, 80000]   
}

df = pd.DataFrame(data)
df
```

```python
# let's count how many we have per country
df.groupby('Country').count()
```

```python
df.groupby('Country')['Age'].mean() # now let's know the average age per country
```

```python
df.groupby('Country')['Salary'].sum() # total sum of their salaries
```

```python
# Multiple Aggregations
df.groupby('Country').agg({'Salary': ['min', 'max'], 'Age': 'mean'})
```

```python
# grouping using more than one column
df.groupby(['Country', 'Age'])['Salary'].count()
```

```python
# Exercise 1

import pandas as pd

# Cargar el archivo CSV
file_path = 'netflix_titles.csv'
netflix_data = pd.read_csv(file_path)

# Vemos si hay algun rating que falte 
missing_ratings = netflix_data['rating'].isnull().sum()
print(f"1. Hay {missing_ratings} calificaciones faltantes.")
```

```python
# Exercise 2

# Definimos el país que queremos analizar (puedes cambiar 'United States' por otro)
country = 'Spain'

# Filtramos los datos para obtener películas de 2021 del país especificado
# Verificamos que el tipo sea 'Movie', el año sea 2021, y el país corresponda al seleccionado
films_2021_country = netflix_data[
    (netflix_data['release_year'] == 2021) & 
    (netflix_data['country'] == country) & 
    (netflix_data['type'] == 'Movie')
].shape[0]  # Contamos el número de filas que cumplen con los filtros

# Imprimimos la cantidad de películas en 2021 de ese país
print(f"2. En 2021, hay {films_2021_country} películas de {country}.")

# En este caso estamos diferenciando el entre "MOVIES" y "TV Shows" ya que si no lo tuvieramos en cuenta nos daria un resultado distinto
```

```python
# Exercise 3.

# Filtramos para seleccionar películas lanzadas en 2020 y que sean del tipo 'Movie'
# Eliminamos las filas que tengan algún valor nulo en las columnas relevantes con 'dropna()'
movies_2020_full_info = netflix_data[
    (netflix_data['release_year'] == 2020) & 
    (netflix_data['type'] == 'Movie')
].dropna().shape[0]  # Contamos las filas resultantes que cumplen todas las condiciones

# Imprimimos la cantidad de películas de 2020 con información completa
print(f"3. En 2020, hay {movies_2020_full_info} películas con información completa.")

```

```python
# Exercise 4

# Contamos cuántos títulos hay por cada año en la columna 'release_year' usando 'value_counts()'
# Obtenemos el año con la mayor cantidad de títulos usando 'idxmax()', que da el índice con el valor más alto
year_with_most_titles = netflix_data['release_year'].value_counts().idxmax()

# Obtenemos la cantidad de títulos para el año con más lanzamientos
most_titles_count = netflix_data['release_year'].value_counts().max()

# Imprimimos el año con más títulos y la cantidad de títulos en ese año
print(f"4. El año con más títulos es {year_with_most_titles} con {most_titles_count} títulos.")

```

```python
# Exercise 5

# Filtramos los datos para obtener solo los lanzamientos desde 2010
# Luego dividimos el total de títulos entre el número de años únicos desde 2010
average_releases_from_2010 = netflix_data[
    netflix_data['release_year'] >= 2010
].shape[0] / len(netflix_data[netflix_data['release_year'] >= 2010]['release_year'].unique())

# Imprimimos el promedio de lanzamientos anuales desde 2010
print(f"5. El promedio anual de lanzamientos desde 2010 es {average_releases_from_2010:.2f}.")


```

```python
# EJERCICIO TITANIC 

# Exercise 1

# Cargar el archivo CSV del Titanic
file_path_titanic = 'train_and_test2.csv'
titanic_data = pd.read_csv(file_path_titanic)

# Cambiar nombre de la columna de supervivencia para evitar confusión
titanic_data.rename(columns={'2urvived': 'Survived'}, inplace=True)

# Calcular el porcentaje de supervivencia por género
gender_survival = titanic_data.groupby('Sex')['Survived'].mean() * 100

# Imprimir el resultado en formato claro
print("1. Porcentaje de supervivencia basado en género:")
for gender, survival_rate in gender_survival.items():
    print(f"  - {gender}: {survival_rate:.2f}%")

```

```python
# Exercise 2

# Calcular el porcentaje de supervivencia agrupado por género y clase
gender_class_survival = titanic_data.groupby(['Sex', 'Pclass'])['Survived'].mean() * 100

# Imprimir el resultado en formato claro
print("\n2. Porcentaje de supervivencia agrupado por género y clase:")
for (gender, pclass), survival_rate in gender_class_survival.items():
    print(f"  - Gender: {gender}, Class: {pclass} => Survival Rate: {survival_rate:.2f}%")
```
